# Task 4: Student Pass/Fail Prediction
### Predict if a student passes or fails based on attendance, homework, and project

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import warnings; warnings.filterwarnings("ignore")
```

```python
np.random.seed(42)
n = 30

# now generate random data for attendance, homework, project, and result in range 0 or 1 (binary)
# p=[0.3,0.7] means 30% chance of 0 and 70% chance of 1 for attendance, similarly for homework and project
attendance = np.random.choice([0,1], n, p=[0.3,0.7])
homework   = np.random.choice([0,1], n, p=[0.35,0.65])
project    = np.random.choice([0,1], n, p=[0.4,0.6])
result     = ((attendance + homework + project) >= 2).astype(int)

df = pd.DataFrame({
    "attendance_above_75": attendance,
    "homework_submitted":  homework,
    "project_completed":   project,
    "result":              result
})
df["result_label"] = df["result"].map({1:"Pass", 0:"Fail"})
print(df.to_string(index=False))
print("\nPass/Fail Distribution:")
print(df["result_label"].value_counts())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
features = ["attendance_above_75","homework_submitted","project_completed"]
labels_map = {0:"No", 1:"Yes"}
for ax, feat in zip(axes, features):
    df.groupby([feat,"result_label"]).size().unstack().plot(kind="bar", ax=ax, edgecolor="black")
    ax.set_title(f"{feat} vs Result", fontweight="bold"); ax.tick_params(rotation=0)
    ax.set_xlabel(feat.replace("_"," ").title())
plt.tight_layout(); plt.savefig("task4_viz.png", dpi=100); plt.show()
```

```python
X = df[features]
y = df["result"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)



# model = DecisionTreeClassifier(random_state=42)
# model.fit(X_train, y_train)
# y_pred = model.predict(X_test)


#use navbias model for classification
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(" Accuracy:", round(accuracy_score(y_test, y_pred)*100, 2), "%")
print("\n Classification Report:")
unique_labels = sorted(set(y_test) | set(y_pred))
label_names = ["Fail","Pass"]
names = [label_names[i] for i in unique_labels]
print(classification_report(y_test, y_pred, labels=unique_labels, target_names=names))

# cm = confusion_matrix(y_test, y_pred)
# plt.figure(figsize=(5,4))
# sns.heatmap(cm, annot=True, fmt="d", cmap="Greens",
#             xticklabels=["Fail","Pass"], yticklabels=["Fail","Pass"])
# plt.title("Confusion Matrix — Pass/Fail", fontweight="bold")
# plt.ylabel("Actual"); plt.xlabel("Predicted")
# plt.tight_layout(); plt.savefig("task4_cm.png", dpi=100); plt.show()
```
